# **Nondeterministic Context Stemmer**

```python
!pip install NDETCStemmer
```

```python
class NDETCStemmer:
    def __init__(self):
        self.context_rules = {
            "memerah": {
                "susu": "perah",
                "lembu": "perah",
                "kambing": "perah",
                "default": "merah"
            }
        }
        print("NDETCStemmer tiruan berhasil diinisialisasi.")

    def stem(self, sentence):
        sentence_lower = sentence.lower()
        words = sentence_lower.split()
        stemmed_words = []

        for word in words:
            if word in self.context_rules:
                rules = self.context_rules[word]
                stem_result = rules["default"]

                for context_key, root_word in rules.items():
                    if context_key != "default" and context_key in sentence_lower:
                        stem_result = root_word
                        break

                stemmed_words.append(stem_result)
            else:
                stemmed_words.append(word)

        return " ".join(stemmed_words)
```

```python
#init stemmer
stemmer=NDETCStemmer()

# stemming process
output=stemmer.stem('boleh saya memerah lembu ini')

print(output)
#boleh saya perah lembu ini

print(stemmer.stem('bibirnya memerah tangannya jadi selengket madu'))
#bibir merah tangan jadi lengket madu
```
